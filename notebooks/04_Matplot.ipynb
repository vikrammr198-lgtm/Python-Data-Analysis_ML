```python
#1 Import Library

import matplotlib.pyplot as plt
import numpy as np
```

```python
#2 Simple Line Plot

x = [1,2,3,4,5]
y = [10,20,30,40,50]

plt.plot(x,y)

plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.show()
```

```python
#3 Line Plot with Markers

x = [1,2,3,4,5]
y = [5,10,15,20,25]

plt.plot(x,y,marker='o')

plt.title("Line Plot with Markers")

plt.show()
```

```python
#4 Multiple Line Plot

x = [1,2,3,4]

y1 = [2,4,6,8]
y2 = [1,3,5,7]

plt.plot(x,y1,label="Series 1")

plt.plot(x,y2,label="Series 2")

plt.legend()

plt.show()
```

```python
#5 Scatter Plot

x = [10,20,30,40,50]
y = [15,25,35,30,60]

plt.scatter(x,y)

plt.title("Scatter Plot")

plt.show()
```

```python
#6 Bar Chart

subjects = ["Python","Java","C++","SQL"]

marks = [90,85,80,88]

plt.bar(subjects,marks)

plt.title("Marks")

plt.show()
```

```python
#7 Horizontal Bar Chart

plt.barh(subjects,marks)

plt.show()
```

```python
#8 Histogram

data = np.random.randint(1,100,100)

plt.hist(data,bins=10,edgecolor='black')

plt.title("Histogram")

plt.show()
```

```python
#9 Pie Chart

languages = ["Python","Java","C"]

students = [40,35,25]

plt.pie(students,labels=languages,autopct="%1.1f%%")

plt.show()
```

```python
#10 Box Plot

marks = [50,60,65,70,75,80,82,90,100]

plt.boxplot(marks)

plt.show()
```

```python
#11 Grid

x = [1,2,3,4]

y = [2,4,6,8]

plt.plot(x,y)

plt.grid()

plt.show()
```

```python
#12 Figure Size

plt.figure(figsize=(8,5))

plt.plot(x,y)

plt.show()
```

```python
#13 Axis Limits

plt.plot(x,y)

plt.xlim(0,5)

plt.ylim(0,10)

plt.show()
```

```python
#14 Save Figure

plt.plot(x,y)

plt.savefig("line_plot.png")

plt.show()
```

```python
#15 Plot Using NumPy

x = np.linspace(0,10,100)

y = np.sin(x)

plt.plot(x,y)

plt.show()
```

```python
#16 Cosine Graph

x = np.linspace(0,10,100)

y = np.cos(x)

plt.plot(x,y)

plt.show()
```

```python
#17 Compare Two Graphs

x = np.linspace(0,10,100)

plt.plot(x,np.sin(x),label="sin(x)")

plt.plot(x,np.cos(x),label="cos(x)")

plt.legend()

plt.show()
```

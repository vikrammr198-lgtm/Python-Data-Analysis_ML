```python
#1 Import Libraries

import pandas as pd
import numpy as np
```

```python
#2 Load Dataset

df = pd.read_csv("automobileEDA.csv")

df.head()
```

```python
#3 Dataset Information

print(df.info())
```

```python
#4 Check Missing Values

print(df.isnull().sum())
```

```python
#5 Display Missing Values

df[df.isnull().any(axis=1)]
```

```python
#6 Fill Missing Values

df["price"].fillna(df["price"].mean(),inplace=True)

df.head()
```

```python
#7 Drop Missing Values

df.dropna(inplace=True)

print(df.shape)
```

```python
#8 Find Duplicate Rows

print(df.duplicated().sum())
```

```python
#9 Remove Duplicate Rows

df.drop_duplicates(inplace=True)

print(df.shape)
```

```python
#10 Rename Columns

df.rename(columns={"horsepower":"HorsePower"}, inplace=True)

print(df.columns)
```

```python
#11 Change Data Type

df["price"] = df["price"].astype(float)

print(df.dtypes)
```

```python
#12 Remove Extra Spaces

df["make"] = df["make"].str.strip()

df.head()
```

```python
#13 Convert Text to Uppercase

df["make"] = df["make"].str.upper()

df.head()
```

```python
#14 Replace Values

df["fuel-type"].replace("gas","petrol", inplace=True)

print(df.head())
```

```python
#15 Detect Outliers (IQR Method)

Q1 = df["price"].quantile(0.25)

Q3 = df["price"].quantile(0.75)

IQR = Q3 - Q1

outliers = df[(df["price"] < (Q1 - 1.5*IQR)) |
              (df["price"] > (Q3 + 1.5*IQR))]

outliers
```

```python
#16 Remove Outliers

df = df[(df["price"] >= (Q1 - 1.5*IQR)) &(df["price"] <= (Q3 + 1.5*IQR))]
```

```python
#17 Check Unique Values

print(df["make"].unique())
```

```python
#18 Value Counts

print(df["make"].value_counts())
```

```python
#19 Sort Dataset

df.sort_values(by="price", ascending=False, inplace=True)

df.head()
```

```python
#20 Reset Index

df.reset_index(drop=True, inplace=True)

df.head()
```

```python
#21 Save Clean Dataset

df.to_csv("cleaned_automobile.csv", index=False)
```

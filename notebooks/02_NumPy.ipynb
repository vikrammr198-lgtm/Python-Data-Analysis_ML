```python
#1 Import NumPy

import numpy as np
```

```python
#2

a = np.array([10, 20, 30, 40, 50])

print(a)
print(type(a))
```

```python
#3 One-Dimensional and Two-Dimensional Arrays

arr1 = np.array([1,2,3,4])

arr2 = np.array([[1,2],
                 [3,4]])

print("1D Array:\n",arr1)
print("2D Array:\n",arr2)
```

```python
#4 Array Properties

print(arr2.ndim)
print(arr2.shape)
print(arr2.size)
print(arr2.dtype)
```

```python
#5 Special Arrays

print(np.zeros((2,3)))

print("\n",np.ones((3,3)))

print("\n",np.eye(4))
```

```python
#6 Range Functions

print(np.arange(1,11))

print("\n",np.linspace(0,1,5))
```

```python
#7 Array Indexing

a = np.array([10,20,30,40,50])

print(a[0])
print(a[-1])
```

```python
#8 Array Slicing

print(a[1:4])

print(a[:3])

print(a[::2])
```

```python
#9 Mathematical Operations

a = np.array([1,2,3])

b = np.array([4,5,6])

print(a+b)

print(a-b)

print(a*b)

print(a/b)
```

```python
#10 Universal Functions

a = np.array([1,4,9,16])

print(np.sqrt(a))

print(np.square(a))

print(np.sin(a))
```

```python
#11 Statistical Functions

a = np.array([10,20,30,40,50])

print(np.mean(a))

print(np.median(a))

print(np.max(a))

print(np.min(a))

print(np.sum(a))

print(np.std(a))
```

```python
#12 Reshape

a = np.arange(12)

print(a)

print(a.reshape(3,4))
```

```python
#13 Flatten

a = np.array([[1,2],[3,4]])

print(a.flatten())
```

```python
#14 Concatenate Arrays

a = np.array([1,2,3])

b = np.array([4,5,6])

print(np.concatenate((a,b)))
```

```python
#15 Stack Arrays

a = np.array([1,2])

b = np.array([3,4])

print(np.vstack((a,b)))

print("\n",np.hstack((a,b)))
```

```python
#16 Boolean Filtering

a = np.array([5,10,15,20,25])

print(a[a>15])
```

```python
#17 Copy vs View

a = np.array([1,2,3])

b = a.copy()

b[0] = 100

print(a)

print(b)
```

```python
#18 Random Numbers

print(np.random.rand(3))

print(np.random.randint(1,100,5))
```

```python
#19 Save and Load Arrays

a = np.array([10,20,30])

np.save("sample.npy",a)

b = np.load("sample.npy")

print(b)
```

```python
#1 Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#2 Load Dataset

df = pd.read_csv("cleaned_automobile.csv")

df.head()
```

```python
#3 Dataset Shape

print("Rows and Columns:", df.shape)
```

```python
#4 Dataset Information

print(df.info())
```

```python
#5 Statistical Summary

df.describe()
```

```python
#6 Check Missing Values

print(df.isnull().sum())
```

```python
#7 Check duplicate values

print(df.duplicated().sum())
```

```python
#8 Correlation Between Numerical Columns

correlation = df.corr(numeric_only=True)

correlation
```

```python
#9 Distribution of Price

plt.hist(df["price"], bins=15,edgecolor='black')

plt.title("Price Distribution")
plt.xlabel("Price")
plt.ylabel("Count")

plt.show()
```

```python
#10 Horsepower vs Price

plt.scatter(df["HorsePower"], df["price"])

plt.title("HorsePower vs Price")
plt.xlabel("HorsePower")
plt.ylabel("Price")

plt.show()
```

```python
#11 Average Price by Car Make

avg_price = df.groupby("make")["price"].mean()

print(avg_price)
```

```python
#12 Bar Chart of Average Price

avg_price.plot(kind="bar")

plt.title("Average Price by Make")
plt.ylabel("Price")

plt.show()
```

```python
#13 Number of Cars by Body Style

body_count = df["body-style"].value_counts()

print(body_count)
```

```python
#14 Bar Chart for Body Style

body_count.plot(kind="bar",edgecolor='black')

plt.title("Body Style Count")

plt.show()
```

```python
#15 Number of Cars by Drive Wheels

print(df["drive-wheels"].value_counts())
```

```python
#16 Pie Chart

df["drive-wheels"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.ylabel("")

plt.show()
```

```python
#17 Horsepower Distribution

plt.hist(df["HorsePower"], bins=10,edgecolor='black')

plt.title("HorsePower Distribution")

plt.show()
```

```python
#18 Top 10 Most Expensive Cars

df.nlargest(10, "price")
```

```python
#19 Cheapest Cars

df.nsmallest(10, "price")
```

```python
#20 Average Mileage

print("In City:",df["city-mpg"].mean())

print("In Highway:",df["highway-mpg"].mean())
```

```python
#21 Box Plot of Price

plt.boxplot(df["price"])

plt.title("Price Box Plot")

plt.show()
```

```python
#22 Relationship Between Wheel Base and Price

plt.scatter(df["wheel-base"], df["price"])

plt.xlabel("Wheel Base")
plt.ylabel("Price")

plt.show()
```

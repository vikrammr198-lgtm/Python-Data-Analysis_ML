```python
#1 Variables and Data Types

name = "Vikram"
age = 21
height = 5.9
is_student = True

print(name)
print(age)
print(height)
print(is_student)

print(type(name))
print(type(age))
print(type(height))
print(type(is_student))
```

```python
#2 User Input

name = input("Enter your name: ")
age = int(input("Enter age: "))

print("Hello", name)
print("Age:", age)
```

```python
#3 Operators

a = 20
b = 6

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
print(a**2)
```

```python
#4 Conditional Statements

marks = 78

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
else:
    print("Grade C")
```

```python
#5 Loops
  #For Loop

for i in range(1,6):
    print(i)
```

```python
#While Loop

i = 1
while i <= 5:
    print(i)
    i += 1
```

```python
#6 Functions

def square(x):
    return x*x

print(square(8))
```

```python
#7 Lists

numbers = [10,20,30,40]

print(numbers)

numbers.append(50)

print(numbers)
print(numbers[2])
```

```python
#7 Tuples

colors = ("Red","Blue","Green")

print(colors)
print(colors[1])
```

```python
#8 Sets

fruits = {"Apple","Orange","Apple"}

print(fruits)
```

```python
#9 Dictionary

student = {
    "Name":"Vikram",
    "Age":21,
    "Branch":"CSE"
}

print(student)
print(student["Name"])
```

```python
#10 Strings

text = "Python"

print(text.upper())
print(text.lower())
print(text[::-1])
```

```python
#11 Exception Handling

try:
    num = int(input("Enter number: "))
    print(100/num)

except ZeroDivisionError:
    print("Cannot divide by zero")

except ValueError:
    print("Invalid input")
```

```python
#12 File Handling

with open("sample.txt","w") as file:
    file.write("Hello Python")

with open("sample.txt","r") as file:
    print(file.read())
```

```python
#13 Modules

import math

print(math.sqrt(81))
print(math.factorial(5))
```

```python
#14 List Comprehension

squares = [i*i for i in range(1,11)]

print(squares)
```

```python
#15 Lambda Function

square = lambda x:x*x
print(square(10))
```

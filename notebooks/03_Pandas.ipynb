```python
#1 Import Pandas

import pandas as pd
```

```python
#2 Create a Series

data = pd.Series([10,20,30,40,50])

print(data)
```

```python
#3 Create a DataFrame

student = {
    "Name":["Ram","Shyam","Anu"],
    "Age":[20,21,19],
    "Marks":[85,90,78]
}

df = pd.DataFrame(student)
df
```

```python
#4 Read CSV File

df = pd.read_csv("Iris.csv")
print(df[1:6])
```

```python
#4 Display First Five records

df.head()
```

```python
#5 Display Last Five Records

df.tail()
```

```python
#6 Dataset Information

print(df.info())
```

```python
#7 Statistical Summary

print(df.describe())
```

```python
#8 Display Columns

print(df.columns)
```

```python
#9 Select Columns

print(df["SepalLengthCm"])

print("\n",df[["SepalLengthCm","PetalLengthCm"]])
```

```python
#10 Select Rows

print(df.iloc[0])

print("\n",df.iloc[0:5])
```

```python
#11 Filter Data

print(df[df["Species"]=="setosa"])
```

```python
#12 Sorting

df.sort_values(by="SepalLengthCm")
```

```python
#13 Add New Column

df["Total"] = df["SepalLengthCm"] + df["PetalLengthCm"]

df.head()
```

```python
#14 Delete Column

df.drop("Total",axis=1,inplace=True)

df.head()
```

```python
#15 Missing Values

print(df.isnull().sum())
```

```python
#16 Fill Missing Values

df.fillna(0,inplace=True)
```

```python
#17 Remove Missing Values

df.dropna(inplace=True)
```

```python
#18 Unique Values

print(df["Species"].unique())
```

```python
#19 Value Counts

print(df["Species"].value_counts())
```

```python
#20 Group By

df.groupby("Species").mean(numeric_only=True)
```

```python
#21 Rename Columns

df.rename(columns={"SepalLengthCm":"Sepal_Length"},inplace=True)

df.head()
```

```python
#22 Export CSV

df.to_csv("iris_output.csv",index=False)
```
